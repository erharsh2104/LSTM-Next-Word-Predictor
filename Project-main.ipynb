```python
import  pandas as pd 
```

```python
data=pd.read_json("News_Category_Dataset_v3.json",lines=True)
```

```python
data.head()
```

```python
data.shape
```

```python
data.columns
```

```python
data=data['short_description'].sample(20000,random_state=42)
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
data.head()
```

```python
data.shape
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
tokenizer=Tokenizer(oov_token="OOV")
```

```python
tokenizer.fit_on_texts(data)
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
data=tokenizer.texts_to_sequences(data)
```

```python
data
```

```python
data[0]
```

```python
input_sentances=[]
for sent in data:
    if len(sent)<2:
        continue
    for i in range(1,len(sent)):
        ngram=sent[:i+1]
        input_sentances.append(ngram)
```

```python
input_sentances
```

```python
max_len=max([len(i) for i in input_sentances])
```

```python
max_len
```

```python
input=pad_sequences(input_sentances,padding='pre',maxlen=max_len)
```

```python
input
```

```python
input[1]
```

```python
input.shape
```

```python
X=input[:,:-1]
```

```python
X.shape
```

```python
Y=input[:,-1]
```

```python
Y
```

```python
Y.shape
```

```python
## Train Test split
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train.shape
```

```python
max_len=X.shape[1]
max_len
```

```python
vocab_size=len(tokenizer.word_index)+1
```

```python
vocab_size
```

```python
model=Sequential([Embedding(vocab_size,128,input_length=max_len),LSTM(256),Dense(vocab_size,activation='softmax')])
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.build(input_shape=(None,max_len))
```

```python
model.summary()
```

```python
model.fit(X_train,Y_train,validation_data=(X_test,Y_test),epochs=10)
```

```python
model.save('Next_word_pred.h5')
```
